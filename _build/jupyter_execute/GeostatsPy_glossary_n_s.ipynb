#### **Normal Score Transformation**

A normal score transform is a [distribution transformation](GeostatsPy_glossary_a_d.ipynb#distribution-transformation) that converts a feature to follow a standard [Gaussian distribution](GeostatsPy_glossary_e_m.ipynb#gaussian-distribution) while preserving the rank order of the data.

The transformation is performed by mapping cumulative probabilities to the inverse Gaussian cumulative distribution function:

$$
Y=\Phi^{-1}(F_X(X))
$$

where:

* $X$ is the original feature

* $F_X$ is the cumulative distribution function of $X$

* $\Phi^{-1}$ is the inverse cumulative distribution function of the standard Gaussian distribution

* $Y$ is the normal score transformed feature

Normal score transforms are commonly applied before Gaussian-based methods such as kriging and sequential Gaussian simulation.

Also see:
- [Distribution Transformation](GeostatsPy_glossary_a_d.ipynb#distribution-transformation)
- [Gaussian Distribution](GeostatsPy_glossary_e_m.ipynb#gaussian-distribution)


#### **Normalized Histogram**

A bar chart of the [univariate](GeostatsPy_glossary_t_z.ipynb#univariate) [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) with probability over an exhaustive set of bins over the range of possible values. These are the steps to build a normalized [histogram](GeostatsPy_glossary_e_m.ipynb#histogram),

1. Divide the [continuous feature](GeostatsPy_glossary_a_d.ipynb#continuous-feature) range of possible values into $K$ equal size bins, $\delta x$:

$$
\Delta x = \left( \frac{x_{max} - x_{min}}{K} \right)
$$

or use available categories for [categorical feature](GeostatsPy_glossary_a_d.ipynb#categorical-feature)s.

2. Count the number of [sample](GeostatsPy_glossary_n_s.ipynb#sample)s (frequency) in each bin, $n_k$, $\forall k=1,\ldots,K$
  
3. Divide each by the total number of data, $n$, to calculate the probability of each bin,

$$
p_k = \frac{n_k}{n}, \forall \quad k = 1,\ldots,L
$$

4. Plot the probability vs. the bin label (use bin centroid if continuous)

Additional comments:

* step 3 converts a standard histogram to a normalized histogram with a y-axis of probability instead of frequency

* for categorical features, a normalized histogram represents an empirical [probability mass function](GeostatsPy_glossary_n_s.ipynb#probability-mass-function)

* for continuous features, the normalized histogram represents the empirical probability of each bin interval and is an approximation to the underlying continuous distribution

Used in: 
- [Normalized Histogram Description and Demonstration](GeostatsPy_univariate_distributions.ipynb#normalized-histogram)

Also see: 
- [Histogram](GeostatsPy_glossary_e_m.ipynb#histogram)


#### **Nugget Effect**

The discontinuity in the [variogram](GeostatsPy_glossary_t_z.ipynb#variogram) at distances smaller than the minimum sampling spacing, representing variability that cannot be resolved by the available sampling.

The nugget effect may result from,

* measurement error

* spatial variability occurring at scales smaller than the sampling interval

It is often communicated as the ratio of the nugget effect to the [sill](GeostatsPy_glossary_n_s.ipynb#sill), known as the relative nugget effect. For example,

* a copper grade [variogram model](GeostatsPy_glossary_t_z.ipynb#variogram-model) with a 20% relative nugget effect

The nugget effect is both,

* a structure interpreted from the [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram)

* a positive-definite variogram structure that may be included as one of the nested structures in a [variogram model](GeostatsPy_glossary_t_z.ipynb#variogram-model)

A large relative nugget effect indicates weak short- [spatial continuity](GeostatsPy_glossary_n_s.ipynb#spatial-continuity) [range](GeostatsPy_glossary_n_s.ipynb#range), while a small relative nugget effect indicates strong local continuity.

#### **Null Hypothesis**

The null hypothesis is the default assumption applied in hypothesis testing that there is no effect, relationship, or statistically significant difference. It is denoted by,

$$
H_0.
$$

Examples include:

$$
H_0:\mu_1 = \mu_2,
$$

or

$$
H_0:\mu_1 \le \mu_2,
$$

where the population means for sample groups 1 and 2 are assumed to be equal.

The alternative hypothesis may be:

* two-sided, testing for any statistically significant difference

* one-sided, testing whether the population parameter is greater than or less than a specified value

Sample data are used to determine whether there is sufficient evidence to reject the null hypothesis.

Failure to reject the null hypothesis does not prove that it is true.

* It indicates only that there is insufficient evidence to conclude otherwise, and the null hypothesis lives on to be tested another day.

Rejecting the null hypothesis does not prove that it is false.

* It indicates that the observed data would be unlikely if the null hypothesis were true, providing evidence in support of the [alternative hypothesis](GeostatsPy_glossary_a_d.ipynb#alternative-hypothesis).

#### **Object-based Simulation**

A stochastic [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) approach that generates [realization](GeostatsPy_glossary_n_s.ipynb#realization)s by sequentially placing parameterized geological objects within the simulation domain. The objects are designed to represent geological shapes observed in nature while allowing variability through distributions of geometric parameters.

The general object-based simulation workflow is:

1. Specify the object templates along with their occurrence probabilities and geometric parameter distributions.

2. Initialize the model with a background [facies](GeostatsPy_glossary_e_m.ipynb#facies).

3. Monte Carlo simulate an object realization, including the object type and its geometric parameters from the specified distributions.

4. Randomly place the geological object within the simulation grid. All grid cells falling inside the object's geometry are assigned the object's facies.

5. Repeat the placement process (return to step 3) until a target facies proportion, object count, or other stopping criterion is reached.

By repeatedly placing stochastic geological objects, realizations reproduce expected geological geometries while representing uncertainty in object occurrence, size, shape, orientation, and location.

* conditioning to local data and representing spatial non-[stationarity](GeostatsPy_glossary_n_s.ipynb#stationarity) in object occurrence and geometry are often the most challenging aspects of object-based simulation

Used in:

#### **Order Relations Correction**

A correction applied to the local distributions estimated by [indicator kriging](GeostatsPy_glossary_e_m.ipynb#indicator-kriging) to ensure they satisfy the axioms of probability.

Order relations correction is required for either,

* [continuous feature](GeostatsPy_glossary_a_d.ipynb#continuous-feature)s - correction of the local [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)

* [categorical feature](GeostatsPy_glossary_a_d.ipynb#categorical-feature)s - correction of the local [probability mass function](GeostatsPy_glossary_n_s.ipynb#probability-mass-function)

Because each threshold or category is estimated independently, the resulting local probability model may not be valid.

For continuous features,

* the estimated cumulative distribution function may not be monotonic increasing, resulting in negative probabilities over portions of the distribution

For categorical features,

* the estimated category probabilities may not sum to one, violating [probability closure](GeostatsPy_glossary_n_s.ipynb#probability-closure)

For continuous features, the order relations correction is,

* a two-pass algorithm that constructs two monotonic cumulative distribution functions and averages them to obtain the corrected cumulative distribution function

For categorical features, the order relations correction is equivalent to the L1 normalization commonly used in machine learning feature engineering,

$$
i(\mathbf{u}_{\alpha};z_k)^{\prime} = \frac{i(\mathbf{u}_{\alpha};z_k)} {\sum_{i=1}^{K} i(\mathbf{u}_{\alpha};z_i)}
$$

where the estimated probability for each category is divided by the sum of all estimated category probabilities so that the corrected probabilities sum to one.

Used in:
- [Indicator Kriging Description and Demonstration](GeostatsPy_categorical_indicator_kriging.ipynb#indicator-kriging-method)
- [Indicator Simulation Description and Demonstration](GeostatsPy_categorical_indicator_simulation.ipynb#sequential-indicator-simulation)



#### **p-value**

A p-value is the probability of observing a test statistic at least as extreme as the calculated value, assuming the [null hypothesis](GeostatsPy_glossary_n_s.ipynb#null-hypothesis) is true.

The p-value is calculated from the appropriate [sampling distribution](GeostatsPy_glossary_n_s.ipynb#sampling-distribution):

$$
p=P(T\geq t_{observed}|H_0)
$$

where:

* $T$ is the test statistic

* $t_{observed}$ is the observed test statistic

* $H_0$ is the null hypothesis

A small p-value indicates that the observed result would be unlikely if the null hypothesis were true.

A hypothesis test typically compares the p-value to the selected [significance level](GeostatsPy_glossary_n_s.ipynb#significance-level):

* if $p \leq \alpha$, reject the null hypothesis

* if $p > \alpha$, fail to reject the null hypothesis

The p-value does not represent:

* the probability that the null hypothesis is true

* the probability that the observed result occurred by chance

#### **Parameter**

A numerical quantity that describes a population or probability model.

Examples include,

* [population](GeostatsPy_glossary_n_s.ipynb#population) [average](GeostatsPy_glossary_a_d.ipynb#average)

#### **Percentile**

A percentile is a value that divides a [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) such that a specified proportion of observations are less than or equal to that value.

The $p$ percentile is calculated from the inverse [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function):

$$
X_p=F_X^{-1}(p)
$$

where:

* $X_p$ is the value at percentile $p$

* $F_X^{-1}$ is the inverse cumulative distribution function

* $p$ is the cumulative probability, where:

$$
0 \leq p \leq 1
$$

For example:

* $X_{0.50}$ is the 50th percentile, also called the [median](GeostatsPy_glossary_e_m.ipynb#median)

* $X_{0.90}$ is the 90th percentile, where 90% of values are less than or equal to this value

Percentiles are commonly used to describe distributions without assuming a particular [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution).

Common percentile summaries include:

* $P10$ - 10th percentile

* $P50$ - 50th percentile or median

* $P90$ - 90th percentile

In uncertainty analysis and [post-processing realizations](GeostatsPy_glossary_n_s.ipynb#post-processing-realizations), percentiles summarize the local distribution of simulated values:

$$
Z_p(\mathbf{u}_{\alpha})=F^{-1}_{Z(\mathbf{u}_{\alpha})}(p)
$$

where the percentile is calculated over an ensemble of [realizations](GeostatsPy_glossary_n_s.ipynb#realization)s at location $\mathbf{u}_{\alpha}$.

Percentiles are useful because they:

* quantify uncertainty without assuming Gaussian behavior

* describe skewed and non-symmetric distributions

* identify extreme or threshold values for decision making

#### **Permeability**

Permeability is the ability of a porous medium to transmit fluids through interconnected pore space.

Permeability is commonly estimated from Darcy's Law:

$$
q=-\frac{kA}{\mu}\frac{dP}{dL}
$$

where:
* $q$ is volumetric flow rate

* $k$ is permeability

* $A$ is cross-sectional flow area

* $\mu$ is fluid viscosity

* $\frac{dP}{dL}$ is the pressure gradient

Permeability is commonly measured in darcies (D) or millidarcies (mD).

Unlike [porosity](GeostatsPy_glossary_n_s.ipynb#porosity), permeability depends strongly on pore connectivity, pore throat geometry, and fractures.

#### **Polygonal Declustering**

A [declustering](GeostatsPy_glossary_a_d.ipynb#declustering) method to assign weights to spatial [sample](GeostatsPy_glossary_n_s.ipynb#sample)s based on local sampling density, such that the weighted statistics are likely more representative of the [population](GeostatsPy_glossary_n_s.ipynb#population). Data weights are assigned so that,

* samples in densely sampled areas receive less weight

* samples in sparsely sampled areas receive more weight

Polygonal declustering proceeds as follows:

1. Split up the [area of interest](GeostatsPy_glossary_a_d.ipynb#area-of-interest) with Voronoi polygons. These are constructed by intersected perpendicular bisectors between adjacent data points. The polygons group the area of interest by nearest data point

2. Assign weight to each datum proportional to the area of the associated Voronoi polygon

$$
w(\bf{u}_j) = n \cdot \frac{A_j}{\sum_{j=1}^n}
$$

where $w(\bf{u}_j)$ is the weight for the $j$ data. Note, the sum of the weights is $n$; therefore, $w(\bf{u}_j)$ is nominal weight of 1.0, sample density if the data were equally spaced over the area of interest.

Here are some highlights for polygonal declustering,

* polygonal declustering is sensitive to the boundaries of the area of interest; therefore, the weights assigned to the data near the boundary of the area of interest may change radically as the area of interest is expanded or contracted

* polygonal declustering is the same as the Theissen polygon method for calculation of precipitation averages developed by Afred H. Thiessen in 1911, {cite}`thiessen1911`

Used in:
- [Polygonal Declustering Description and Demonstration](GeostatsPy_declustering_all_methods.ipynb#polygonal-declustering)

Also see:

The distribution is defined by the parameter $\lambda$, where:

$$
\mu=\lambda
$$

$$

The Poisson distribution is commonly used for count data, such as the number of events, occurrences, or objects within a defined area or interval.

#### **Population**

The complete set of values for a [feature](GeostatsPy_glossary_e_m.ipynb#feature) over the 2D [area of interest](GeostatsPy_glossary_a_d.ipynb#area-of-interest) or 3D [volume of interest](GeostatsPy_glossary_t_z.ipynb#volume-of-interest), represented at sufficient resolution to support [decision making](GeostatsPy_glossary_a_d.ipynb#decision-making).

For example,

* the exhaustive set of porosity values at every location within a reservoir

* the exhaustive set of gold grades throughout an ore body

In practice, the entire population is rarely observed. Instead, a limited [sample](GeostatsPy_glossary_n_s.ipynb#sample) is collected and used to infer population [parameters](GeostatsPy_glossary_n_s.ipynb#parameter).

#### **Porosity**

Porosity is the fraction of the total rock volume occupied by pore space available to store fluids.

Porosity is calculated as:

$$
\phi=\frac{V_p}{V_b}
$$

where:

* $\phi$ is porosity (fraction)

* $V_p$ is pore volume

* $V_b$ is bulk rock volume

Porosity ranges from 0 to 1 and is commonly reported as a percentage.

Porosity is a fundamental petrophysical property that controls the storage capacity of a reservoir but does not necessarily indicate its ability to transmit fluids.

#### **Post-Processing Realizations**

Post-processing realizations is the analysis of an ensemble of [realization](GeostatsPy_glossary_n_s.ipynb#realizations)s and [scenarios](GeostatsPy_glossary_n_s.ipynb#scenarios) to summarize spatial uncertainty and support [decision making](GeostatsPy_glossary_a_d.ipynb#decision-making).

A stochastic [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) produces multiple equally probable models of the subsurface:

$$
Z^{(1)}(\mathbf{u}_{\alpha}),Z^{(2)}(\mathbf{u}_{\alpha}),...,Z^{(L)}(\mathbf{u}_{\alpha})
$$

where:

* $Z^{(\ell)}(\mathbf{u}_{\alpha})$ is the simulated value for realization $\ell$ at location $\mathbf{u}_{\alpha}$

* $\ell=1,\ldots,L$ is the realization index

* $\mathbf{u}_{\alpha}$ is a location vector in the model domain

Post-processing summarizes the local uncertainty distribution over the ensemble of realizations.

Common post-processing methods include:

E-type estimate - the E-type estimate is the local [expected value](GeostatsPy_glossary_e_m.ipynb#expectation) calculated as the average over all realizations:

$$
E[Z(\mathbf{u}_{\alpha})]
\approx
\frac{1}{L}
\sum_{\ell=1}^{L}
Z^{(\ell)}(\mathbf{u}_{\alpha})
$$

The E-type estimate provides a single representative model, but may smooth local variability and does not represent a single possible subsurface outcome.

Conditional variance - the local conditional variance measures uncertainty at location $\mathbf{u}_{\alpha}$:

$$
\sigma^2(\mathbf{u}_{\alpha})\approx\frac{1}{L-1}\sum_{\ell=1}^{L}\left(Z^{(\ell)}(\mathbf{u}_{\alpha})-E[Z(\mathbf{u}_{\alpha})]\right)^2
$$

Higher conditional variance indicates greater uncertainty among the realizations.

Local percentiles - percentiles summarize the local uncertainty distribution from the ensemble.

The $p$ percentile is:

$$
Z_p(\mathbf{u}_{\alpha})=F^{-1}_{Z(\mathbf{u}_{\alpha})}(p)
$$

where:

* $p$ is the cumulative probability

* $F^{-1}_{Z(\mathbf{u}_{\alpha})}$ is the inverse cumulative distribution function derived from the ensemble values

Common examples include:

* $P10$ - optimistic high case

* $P50$ - median case

* $P90$ - conservative low case

Probability of exceedance - the probability that a variable exceeds a threshold $z_c$ is:

$$
P(Z(\mathbf{u}_{\alpha})>z_c)\approx\frac{1}{L}\sum_{\ell=1}^{L}I\left(Z^{(\ell)}(\mathbf{u}_{\alpha})>z_c\right)
$$

where $I$ is an indicator function:

$$
I(A)=
\begin{cases}
1,& \text{if } A \text{ is true}\\
0,& \text{otherwise}
\end{cases}
$$

Post-processing is essential because individual realizations represent possible subsurface states, while summary models quantify uncertainty and support decisions.

#### **Posterior**

In [Bayes' Theorem](GeostatsPy_probability.ipynb#bayes-theorem), the posterior probability represents updated knowledge or uncertainty about possible model assumptions, states, or parameters after incorporating new data. The posterior is calculated by combining the [prior](GeostatsPy_glossary_n_s.ipynb#prior) probability with the [likelihood function](GeostatsPy_glossary_e_m.ipynb#likelihood-function) describing the compatibility of observed data with possible models,

$$
P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}
$$

where:

- $P(A)$ is the [prior](GeostatsPy_glossary_n_s.ipynb#prior) probability representing uncertainty before observing new information,

- $P(B|A)$ is the [likelihood function](GeostatsPy_glossary_e_m.ipynb#likelihood-function) describing the compatibility of observations $B$ with state or parameter $A$,

- $P(B)$ is the [evidence](GeostatsPy_glossary_e_m.ipynb#evidence) term used to normalize the posterior probability,

- $P(A|B)$ is the [posterior](GeostatsPy_glossary_n_s.ipynb#posterior) probability representing updated uncertainty after incorporating observations.


#### **Power Law Average**

A flexible family of averaging methods used to scale a [feature](GeostatsPy_glossary_e_m.ipynb#feature) from a smaller [volume support](GeostatsPy_glossary_t_z.ipynb#volume-support), $v$, to a larger support, $V$, by calculating an effective value representative of the larger volume.

The power law average is,

$$
z_V =\left[ \frac{1}{n} \sum_{i=1}^{n} z_{v,i}^{\,\omega} \right]^{\frac{1}{\omega}}
$$

where $\omega$ is the averaging power.

Special cases include,

* $\omega = 1$ — arithmetic [average](GeostatsPy_glossary_a_d.ipynb#average)

* $\omega = -1$ — harmonic average

* $\omega \rightarrow 0$ — geometric average (obtained in the limit)

The choice of averaging power depends on the physical process being modeled. For example, for permeability,

* arithmetic averaging is appropriate for flow parallel to bedding

* harmonic averaging is appropriate for flow perpendicular to bedding

* near-geometric averaging is often appropriate for oblique flow directions

Used in:

#### **PP Plot**

A probability-probability (PP) plot is a graphical method for comparing two [statistical distributions](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) by plotting corresponding cumulative probabilities against each other.

For two distributions $X$ and $Y$, the PP plot displays:

$$
(F_X(x),F_Y(x))
$$

where:

* $F_X(x)$ is the [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function) of distribution $X$ evaluated at value $x$

* $F_Y(x)$ is the cumulative distribution function of distribution $Y$ evaluated at the same value $x$

The PP plot is created by plotting paired cumulative probabilities over the range of observed values:

$$
x \in [x_{min},x_{max}]
$$

If two distributions are identical:

$$
F_Y(x)=F_X(x)
$$

and the PP plot follows the 45-degree line:

$$
y=x
$$

PP-plots are commonly used to:

* compare an empirical [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function) with a modeled distribution

* evaluate agreement between two probability models

* identify differences in probability behavior

Common PP-plot interpretations:

* deviations indicate differences in cumulative probability

* systematic curvature indicates differences in distribution shape

* deviations near the upper or lower ends indicate differences in tails or extreme probabilities

Compared with QQ plots:

* PP plots compare cumulative probabilities using the forward CDF:

$$
x \rightarrow F(x)
$$

* QQ plots compare corresponding values using the inverse CDF:

$$
p \rightarrow F^{-1}(p)
$$

#### **Prediction Interval**

An uncertainty range for a future [prediction](GeostatsPy_glossary_n_s.ipynb#prediction) represented by lower and upper bounds based on a specified probability level, known as the confidence level.

For example, a 95% prediction interval may be communicated as,

* given [predictor feature](GeostatsPy_glossary_n_s.ipynb#predictor-feature) values, $X_1=x_1,\ldots,X_m=x_m$, there is a 95% probability that a future reservoir NTG observation will fall between 13% and 17%.

A prediction interval represents uncertainty in the next observation and integrates,

* uncertainty in the estimated model prediction, $\hat{Y}|X=x$

* irreducible variability or error in the response around the model prediction, $Y-\hat{Y}$

Therefore, prediction intervals are wider than confidence intervals for the estimated [mean](GeostatsPy_glossary_e_m.ipynb#mean) response because they include both model uncertainty and observation uncertainty.

The prediction interval depends on,

* uncertainty in [model parameter](GeostatsPy_glossary_e_m.ipynb#model-parameter)s

* uncertainty in the estimated conditional mean response

* variability of observations around the conditional mean response

#### **Prior**

In [Bayes' Theorem](GeostatsPy_probability.ipynb#bayes-theorem), the prior probability represents knowledge or uncertainty about possible model assumptions, states, or parameters before considering new data. The prior is combined with the [likelihood function](GeostatsPy_glossary_e_m.ipynb#likelihood-function) describing new observations to calculate the [posterior](GeostatsPy_glossary_n_s.ipynb#posterior) probability,

$$
P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}
$$

where:

- $P(A)$ is the [prior](GeostatsPy_glossary_n_s.ipynb#prior) probability representing uncertainty before observing new information,

- $P(B|A)$ is the [likelihood function](GeostatsPy_glossary_e_m.ipynb#likelihood-function) describing the compatibility of observations $B$ with state or parameter $A$,

- $P(B)$ is the [evidence](GeostatsPy_glossary_e_m.ipynb#evidence) term used to normalize the posterior probability,

- $P(A|B)$ is the [posterior](GeostatsPy_glossary_n_s.ipynb#posterior) probability representing updated uncertainty after incorporating observations.


#### **Probability Closure**

The normalization requirement of probability measures stating that the total probability over the entire sample space $\Omega$ is equal to 1:

$$
P(\Omega) = 1
$$

This property ensures that all possible outcomes collectively account for the entire probability mass.

Useful examples include:

* Closure for complements:

$$
P(A) + P(A^c) = 1
$$

* Conditional complements:

$$
P(A|B) + P(A^c|B) = 1
$$

* [Probability density function](GeostatsPy_glossary_n_s.ipynb#probability-density-function)s:

$$
\int_{-\infty}^{\infty} f_X(x)\,dx = 1
$$

* [Probability mass function](GeostatsPy_glossary_n_s.ipynb#probability-mass-function)s:

$$
\sum_{k=1}^{K} p_k = 1
$$

* [Cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)s:

$$
\lim_{x \to \infty} F_X(x) = 1
$$

Used in:
- [Probability Operations Itemization and Demonstration](GeostatsPy_probability.ipynb#constraints-on-probability)
  

#### **Probability Constraints**

The fundamental requirements for valid measures of probability include,

1. Boundedness, probabilities must be between zero and one,

$$
0.0 \le P(A) \le 1.0
$$

2. Closure, the total probability over the entire sample space, $\Omega$, is one,

$$
P(\Omega) = 1.0
$$

3. Null set, the probability of the empty set is zero,

$$
P(\emptyset) = 0.0
$$

4. Additivity, the probability of [mutually exclusive events](GeostatsPy_glossary_e_m.ipynb#mutually-exclusive-events) is the sum of their individual probabilities,

$$
P\left(\bigcup_i A_i\right) = \sum_i P(A_i)
$$

These constraints are closely related to the [Kolmogorov probability axioms](GeostatsPy_glossary_e_m.ipynb#kolmogorov-probability-axioms).

Used in:
- [Probability Operations Itemization and Demonstration](GeostatsPy_probability.ipynb#constraints-on-probability)
  

#### **Probability Density Function**

A representation of a continuous [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) with a density function, $f(x)$, describing the relative density over the range of possible feature values, $x$.

A [univariate](GeostatsPy_glossary_t_z.ipynb#univariate) probability density function is denoted by,

$$
f_X(x)
$$

and a [bivariate](GeostatsPy_glossary_a_d.ipynb#bivariate) probability density function is denoted by,

$$
f_{X,Y}(x,y)
$$

and extends to [multivariate](GeostatsPy_glossary_e_m.ipynb#multivariate) distributions.

For example, the Gaussian probability density function is specified as,

$$
X \sim \mathcal{N}(\mu,\sigma^2)
$$

with probability density function,

$$
f_X(x) = \frac{1}{\sigma\sqrt{2\pi}}\exp \left(-\frac{(x-\mu)^2}{2\sigma^2} \right), \qquad -\infty < x < \infty
$$

parameterized by [average](GeostatsPy_glossary_a_d.ipynb#average), $\mu$, and variance, $\sigma^2$.

These are requirements for a valid probability density function,

* non-negativity constraint, the density cannot be negative,

$$
0.0 \le f_X(x)
$$

* the density value may be greater than 1.0 because density is not probability

* integrate density over a range of $x$ to calculate probability,

$$
0 \le \int_a^b f_X(x) dx = P(a \le x \le b) \le 1.0
$$

* [probability closure](GeostatsPy_glossary_n_s.ipynb#probability-closure), the total area under the PDF curve is equal to 1.0,

$$
\int_{-\infty}^{\infty} f_X(x) dx = 1.0
$$

Nonparametric PDFs are commonly calculated with kernels (usually a small Gaussian distribution) that are summed over all data. Therefore, there is an implicit scale (smoothness) parameter when calculating a PDF.

* too large of kernels will smooth out important information about the univariate distribution

* too narrow a kernel will result in an overly noisy PDF that is difficult to interpret

This is analogous to the choice of bin size for a [histogram](GeostatsPy_glossary_e_m.ipynb#histogram) or [normalized histogram](GeostatsPy_glossary_n_s.ipynb#normalized-histogram).

Parametric PDFs require model fitting to the data. The steps are,

1. Select a parametric distribution, e.g., Gaussian, lognormal, etc.

2. Calculate the parameters for the parametric distribution based on available data, using methods such as least squares or maximum likelihood.

It is very common to use the acronym PDF for probability density function.

Used in:
- [Comparison of Inference and Prediction](GeostatsPy_concepts.ipynb#inference-and-prediction)

Contrast with:
- [Cumulative Distribution Function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)
  

#### **Probability Mass Function**

A function that describes the probability distribution of a [discrete feature](GeostatsPy_glossary_a_d.ipynb#discrete-feature). The probability mass function assigns a probability to each possible discrete outcome,

$$
p_X(x) = P(X=x)
$$

with the requirements,

$$
p_X(x) \geq 0, \quad \forall x
$$

and [probability closure](GeostatsPy_glossary_n_s.ipynb#probability-closure),

$$
\sum_x p_X(x) = 1.0
$$

The [normalized histogram](GeostatsPy_glossary_n_s.ipynb#normalized-histogram) of a discrete feature is an empirical probability mass function.

#### **Probability Operators**

A list of useful, common probability operators that are essential for working with probability and uncertainty problems.

1. [Union of Events](GeostatsPy_glossary_t_z.ipynb#union-of-events) - the union of outcomes, the probability of $A$ or $B$ is calculated with the probability addition rule,

$$
P(A \cup B) = P(A) + P(B) - P(A,B)
$$

$\quad$ where the intersection probability is subtracted to avoid double counting outcomes common to both events.

2. [Intersection of Events](GeostatsPy_glossary_e_m.ipynb#intersection-of-events) - the intersection of outcomes, the probability of $A$ and $B$ is represented as,

$$
P(A \cap B) = P(A,B)
$$

$\quad$ Under the assumption of [independence](GeostatsPy_glossary_e_m.ipynb#independence) of $A$ and $B$, the intersection probability can be calculated from the marginal probabilities,

$$
P(A,B) = P(A) \cdot P(B)
$$

$\quad$ If there is dependence between $A$ and $B$, then [conditional probability](GeostatsPy_glossary_a_d.ipynb#conditional-probability) is required,

$$
P(A,B) = P(A|B) \cdot P(B)
$$

3. [Complementary Events](GeostatsPy_glossary_a_d.ipynb#complementary-events) - the NOT operator for probability. If we define event $A$, then the complement $A^c$ represents all outcomes that are not $A$. 

$\quad$ The resulting closure relationship is,

$$
P(A) + P(A^c) = 1.0
$$

$\quad$ Complementary events extend naturally to conditional probabilities, for example,

$$
P(A|B) + P(A^c|B) = 1.0
$$

$\quad$ Note, the conditioning event must remain the same.

4. [Mutually Exclusive Events](GeostatsPy_glossary_e_m.ipynb#mutually-exclusive-events) - events that do not intersect and have no common outcomes. Using set notation,

$$
\{x: x \in A \text{ and } x \in B \} = \emptyset
$$

$\quad$ and the [joint probability](GeostatsPy_glossary_e_m.ipynb#joint-probability) is,

$$
P(A \cap B) = P(A,B) = 0
$$

Used in:
- [Probability Operations Itemization and Demonstration](GeostatsPy_probability.ipynb#probability-operations)


#### **Probability Perspectives**

The three primary perspectives for interpreting and calculating probability are:

1. Long-term frequencies - probability as the ratio of observed outcomes from repeated experiments. This perspective requires repeatable experiments and observations, and is the basis for [frequentist probability](GeostatsPy_glossary_e_m.ipynb#frequentist-probability).

2. Physical tendencies or propensities - probability based on knowledge of, or models for, the physical system. For example, the probability of a heads outcome from a coin toss can be known from the physical properties of the coin without performing repeated experiments.

3. Degrees of belief - probability representing our uncertainty about an outcome or proposition, allowing probabilities to be updated as new information becomes available. This perspective is the basis for [Bayesian probability](GeostatsPy_glossary_a_d.ipynb#bayesian-probability).

#### **Proportional Effect**

The proportional effect is the tendency for the uncertainty (variance or standard deviation) of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) to increase as its mean or expected value increases. It is a common form of [heteroscedasticity](GeostatsPy_glossary_e_m.ipynb#heteroscedasticity) observed in many geological and engineering variables.

The proportional effect commonly results in positively skewed distributions because larger values have greater absolute variability. It is frequently observed for variables such as permeability, production rate, thickness, and ore grade.

For example, assuming a 10% relative uncertainty:

- for a porosity estimate of 2%, the standard deviation of estimation error is 0.2%

- for a porosity estimate of 20%, the standard deviation of estimation error is 2.0%

Recognition of the proportional effect is important when selecting estimation methods and uncertainty models, as prediction uncertainty should generally increase with the expected value.

#### **Proportions**

The proportion of each possible category relative to the total number of observations. Proportions describe the categorical distribution of a feature and are equivalent to the probability of occurrence when observations are considered representative.

For a [sample](GeostatsPy_glossary_n_s.ipynb#sample), the proportion of category $k$ is,

$$
p_k = \frac{n_k}{n}
$$

where $n_k$ is the number of observations in category $k$ and $n$ is the total number of observations.

The proportions satisfy [probability closure](GeostatsPy_glossary_n_s.ipynb#probability-closure),

$$
\sum_{k=1}^{K} p_k = 1.0
$$

where $K$ is the total number of categories.

Proportions are,

* the y-axis of a [categorical feature](GeostatsPy_glossary_a_d.ipynb#categorical-feature) [probability density function](GeostatsPy_glossary_n_s.ipynb#probability-density-function)

* the [marginal probability](GeostatsPy_glossary_e_m.ipynb#marginal-probability)'s for a categorical feature

* commonly used to summarize lithology, [facies](GeostatsPy_glossary_e_m.ipynb#facies), rock type, mineral class, land use, or any categorical variable

Facies proportions are central to geostatistical modeling because facies often define stationary domains with distinct statistical and spatial characteristics. For example,

* sand — high porosity and permeability occurring in relatively large connected bodies

* shale — low porosity and permeability occurring in drapes and thin, laterally continuous beds

The determination of facies proportions is one of the most important modeling decisions because they control,

* expected volumetrics of each facies

* connectivity and geological architecture

* flow simulation and production forecasts

* uncertainty in downstream decision making

Facies proportions may be estimated from,

* well or drill-hole observations

* interpreted [seismic](GeostatsPy_glossary_n_s.ipynb#seismic) data

* geological analogs

* conceptual geological models

Because available data are sparse, the proportions themselves are uncertain. This uncertainty is commonly represented with multiple [scenarios](GeostatsPy_glossary_n_s.ipynb#scenarios) by varying the global proportions within plausible limits and evaluating the impact on the resulting subsurface models.

#### **QQ Plot**

A quantile-quantile (QQ) plot is a graphical method for comparing two [statistical distributions](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) by plotting corresponding quantiles against each other.

For two distributions $X$ and $Y$, the QQ plot displays:

$$
(F_X^{-1}(p),F_Y^{-1}(p))
$$

where:

* $F_X^{-1}(p)$ is the inverse [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function) of distribution $X$ evaluated at cumulative probability $p$

* $F_Y^{-1}(p)$ is the inverse cumulative distribution function of distribution $Y$ evaluated at cumulative probability $p$

The QQ plot is created by plotting paired quantiles over a range of cumulative probability values:

$$
p \in [0,1]
$$

If two distributions are similar, the QQ plot approximately follows a straight line:

$$
Q_Y(p)=a+bQ_X(p)
$$

where:

* $a$ represents a location shift

* $b$ represents a scale difference

For identical distributions:

$$
Q_Y(p)=Q_X(p)
$$

and the QQ plot follows the 45-degree line:

$$
y=x
$$

QQ-plots are commonly used to:

* compare a sample distribution with a theoretical distribution, such as a [Gaussian distribution](GeostatsPy_glossary_e_m.ipynb#gaussian-distribution)

* identify departures from distributional assumptions

* evaluate normality

Common QQ-plot interpretations:

* shifts indicate differences in location or mean

* curvature indicates differences in distribution shape

* deviations in the tails indicate differences in extreme values

* different slopes indicate differences in scale or variance

Quantitative features may be continuous or discrete depending on whether the possible values are measured along a continuum or occur as countable values.

#### **R Squared**

Stated as coeficient of determination, "R-squared" or in mathatical notations as $R^2$, a statistical measure of the proportion of the variability in a [response feature](GeostatsPy_glossary_n_s.ipynb#response-feature) that is explained by a regression model.

For ordinary least-squares [linear regression](GeostatsPy_glossary_e_m.ipynb#linear-regression) with an intercept, the response feature [variance](GeostatsPy_glossary_t_z.ipynb#variance) can be partitioned into,

$$
\sigma^2_{tot}=\sigma^2_{reg}+\sigma^2_{res}
$$

where,

* $\sigma^2_{tot}$ is the variance of the response feature, $y_i$

* $\sigma^2_{reg}$ is the variance explained by the regression model

* $\sigma^2_{res}$ is the residual variance, variance of prediction error

The coefficient of determination is therefore,

$$
r^2=\frac{\sigma^2_{reg}}{\sigma^2_{tot}}=1-\frac{\sigma^2_{res}}{\sigma^2_{tot}}
$$

where,

* $R^2=0$ indicates the regression explains none of the response variability

* $R^2=1$ indicates the regression explains all of the response variability

For ordinary least-squares linear regression with a single [predictor feature](GeostatsPy_glossary_n_s.ipynb#predictor-feature),

$$
R^2=\rho_{x,y}^2
$$

where $\rho_{x,y}$ is the [correlation coefficient](GeostatsPy_glossary_a_d.ipynb#correlation-coefficient) between the predictor and response.

Some additional comments,

* for many nonlinear regression models an $R^2$ statistic can still be calculated, but it should be interpreted as a relative reduction in squared prediction error rather than a strict variance decomposition

* $R^2$ does not indicate whether predictions are unbiased or whether the model generalizes well to unseen data; prediction error metrics such as mean square error, root mean square error, or mean absolute error are required to evaluate predictive performance.

#### **Random Function**

A set of [random variable](GeostatsPy_glossary_n_s.ipynb#random-variable)s correlated over space or time. In [geostatistics](GeostatsPy_glossary_e_m.ipynb#geostatistics), a random function provides the mathematical framework for representing spatial uncertainty and variability.

The key concept introduced by Matheron is that a geological phenomenon is viewed as a [realization](GeostatsPy_glossary_n_s.ipynb#realization) of an underlying random function. The observed spatial data represent one possible outcome from this random process.

Important points on random function nomenclature,

* random variables are denoted with upper-case, e.g., $X$

* random functions are denoted with upper-case with location vectors, e.g.,

$$
X(\mathbf{u}_1), X(\mathbf{u}_2), \ldots, X(\mathbf{u}_n)
$$

* joint outcomes called realizations, or data [sample](GeostatsPy_glossary_n_s.ipynb#sample)s are represented with lower case, e.g.,

$$
x(\mathbf{u}_1), x(\mathbf{u}_2), \ldots, x(\mathbf{u}_n)
$$

* realizations with the $\ell$ notation, e.g.,

$$
x^{\ell}(\mathbf{u}_1), x^{\ell}(\mathbf{u}_2), \ldots, x^{\ell}(\mathbf{u}_n)
$$

for $\ell = 1,\ldots,L$ realizations.

- [Subsurface Uncertainty Modeling Introductory Discussion](GeostatsPy_concepts.ipynb#subsurface-uncertainty-modeling)


#### **Random Sampling**

A [representative spatial sampling](GeostatsPy_glossary_n_s.ipynb#representative-spatial-sampling) method from a:

- population – sample for statistical inference
- data set – bootstrap sample generated by sampling with replacement
- model – stochastic realization

Random sampling requires that:

- every possible sample of the specified size is equally likely to be selected
- each selection is made randomly and is not influenced by previous selections or outcomes

This minimizes selection bias and supports valid statistical inference.

#### **Random Variable**

A mathematical representation of uncertainty where the value of a feature is unknown and can take a range of possible outcomes described by a [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution), [probability density function](GeostatsPy_glossary_n_s.ipynb#probability-density-function), or [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function).

A random variable is denoted with upper-case notation, e.g., $X$, while possible outcomes or observed values are represented with lower-case notation, e.g., $x_{\alpha}$ or [realization](GeostatsPy_glossary_n_s.ipynb#realization) $x^{\ell}$.

For spatial phenomena, a location vector, $\mathbf{u}$, is added to represent the random variable at a specific location,

* spatial random variable:

X(\mathbf{u}_{\alpha})
$$

* spatial data measure:

$$
x(\mathbf{u}_{\alpha})
$$

* spatial realization:

$$
x^{\ell}(\mathbf{u}_{\alpha})
$$

The collection of correlated spatial random variables over many locations forms a [random function](GeostatsPy_glossary_n_s.ipynb#random-function).

#### **Range**

There are two common uses for the term range:

1. statistical range as a measure of dispersion for a distribution

2. variogram range as a parameter describing spatial continuity

These two uses describe different concepts and should be interpreted from context.

The statistical range is a measure of dispersion (spread) of a [univariate](GeostatsPy_glossary_t_z.ipynb#univariate) distribution. It is calculated as:

$$
X_{range}=X_{max}-X_{min}
$$

or using the [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function):

$$
X_{range}=F_X^{-1}(1.0)-F_X^{-1}(0.0)
$$

where $F_X^{-1}$ is the inverse cumulative distribution function (quantile function) for feature $X$.

The variogram range is the [lag distance](GeostatsPy_glossary_e_m.ipynb#lag-distance) where the [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram) approximately reaches the [sill](GeostatsPy_glossary_n_s.ipynb#sill), indicating the distance beyond which there is no additional modeled [spatial continuity](GeostatsPy_glossary_n_s.ipynb#spatial-continuity).

Important aspects of variogram range include:

* for lag distances less than the range, nearby [samples](GeostatsPy_glossary_n_s.ipynb#sample) provide information about each other through spatial continuity.

* for lag distances at and beyond the range, the variogram model indicates no additional spatial continuity.

* range is a [variogram model](GeostatsPy_glossary_t_z.ipynb#variogram-model) parameter used to define permissible, positive definite [variogram models](GeostatsPy_glossary_t_z.ipynb#variogram-model).

* for nested variogram models, each spatial structure may have its own range.

* when the range varies by direction, this is called [geometric anisotropy](GeostatsPy_glossary_e_m.ipynb#geometric-anisotropy).

#### **Representative Spatial Sampling**

The [sample](GeostatsPy_glossary_n_s.ipynb#sample) and resulting sample statistics are representative of the [population](GeostatsPy_glossary_n_s.ipynb#population), by sampling theory we have 2 options:

1. Random sampling - each potential sample from the population is equally likely to be sampled as samples are collected. This includes,

* selecting a specific location has no impact on the selection of subsequent locations.

* assumption that the population size that is much larger than the sample size; therefore, significant correlation between samples is not imposed due to without replacement sampling (the constraint that you can only sample a location once). Note, generally this is not an issue for the subsurface due to the sparsely sampled massive populations

2. Regular sampling - sampling at equal space or time intervals. While random sampling is prefered, regular sampling is robust as long as,

* the regular sampling intervals do not align with natural periodicity in the data, e.g., the crests are systemally sampling resulting in biased high sample statistics

#### **Residual**

A residual is the difference between an observed value and a model prediction or estimate.

The residual is calculated as:

$$
r_i=y_i-\hat{y}_i
$$

where:

* $r_i$ is the residual for observation $i$

* $y_i$ is the observed value

* $\hat{y}_i$ is the predicted or estimated value

Residuals represent the portion of the observed data not explained by the model.

For a well-performing model:

* residuals should have approximately zero mean

$$
E[r]\approx 0
$$

* residuals should not show systematic patterns

* residual variability should be consistent with the assumed uncertainty model

Residual analysis is used to evaluate:

* model assumptions

* [bias](GeostatsPy_glossary_a_d.ipynb#bias)

* unexplained variability

* spatial continuity of remaining variation

In spatial modeling, residuals may be analyzed using:

* residual [histograms](GeostatsPy_glossary_e_m.ipynb#histogram)

* residual [variograms](GeostatsPy_glossary_t_z.ipynb#variogram)

* residual maps

The relationship between error and residual is:

* error is the difference from the unknown true value:

$$
e=\hat{Z}-Z
$$

* residual is the measurable difference between observed and predicted values:

$$
r=y-\hat{y}
$$

Residuals are commonly used in:

* [linear regression](GeostatsPy_glossary_e_m.ipynb#linear-regression)

* [loss functions](GeostatsPy_glossary_e_m.ipynb#loss-function)

* model validation

* [estimation](GeostatsPy_glossary_e_m.ipynb#estimation)

#### **Resolution**

Resolution is the smallest spatial scale at which a measurement or model can distinguish meaningful variation. 

* achievable resolution is limited by the physics of the measurement system, while the selected model resolution is often a practical choice based on modeling objectives, data availability, computational resources, and storage constraints.

* [volume support](GeostatsPy_glossary_t_z.ipynb#volume-support) describes the physical volume represented by an individual measurement or model value. Resolution and volume support are related, but resolution may be limited by measurement physics even when smaller volume supports are selected.

Three basic aspects of spatial data include:

1. resolution
2. [coverage](GeostatsPy_glossary_a_d.ipynb#coverage)
3. [property](GeostatsPy_glossary_n_s.ipynb#property) or information

Examples of data resolution include,

* [core](GeostatsPy_glossary_a_d.ipynb#core) – scanned core provides geological information, including sedimentary structures, textures, fractures, and mineralogy, at camera resolutions of approximately 100–250 μm/pixel.

* [well log](GeostatsPy_glossary_t_z.ipynb#well-log) – conventional wireline gamma ray measurements are vertically averaged over approximately 30–60 cm (1–2 ft) along the borehole.

* [seismic](GeostatsPy_glossary_n_s.ipynb#seismic) – deepwater Gulf of Mexico inverted seismic acoustic impedance volumes typically have a dominant frequency of 20–50 Hz and P-wave velocities of 2,500–4,000 m/s, giving a quarter-wavelength vertical resolution of approximately **10–25 m**. Model-based inversion constrained by well logs estimates acoustic impedance with an effective vertical resolution of approximately **5–15 m**.

An example of model resolution includes,

* Based on the modeling objectives and available computational and storage resources, a subsurface model is discretized into 50,000,000 cells (1,000 × 1,000 × 50). To represent the reservoir, each model cell has dimensions of 100 × 100 × 2 m, which defines the model resolution.

Resolution spans many orders of magnitude in subsurface characterization. For example, core images may resolve features on the order of 100 μm, well logs average over 30–60 cm, while inverted seismic resolves features on the order of 5–15 m. Successful subsurface modeling integrates information across these multiple scales.

There is often a resolution–coverage trade-off with spatial data:

* High-resolution data sources tend to have limited coverage – core data provides measurements at micrometer to millimeter scales, but only samples a very small fraction of the reservoir volume. A core may represent only approximately one ten-billionth to one hundred-billionth of the reservoir volume.

* High-coverage data sources tend to have lower resolution – remote sensing and seismic data may cover the entire reservoir or field area, but typically represent geological properties at resolutions of tens to hundreds of meters.

#### **Rock Quality Index**

The Rock Quality Index (RQI) is a petrophysical measure of rock quality that combines [permeability](GeostatsPy_glossary_n_s.ipynb#permeability) and [porosity](GeostatsPy_glossary_n_s.ipynb#porosity) to quantify the effectiveness of pore networks for fluid flow.

The Rock Quality Index is calculated as:

$$
RQI = 0.0314\sqrt{\frac{k}{\phi}}
$$

where:

* $k$ is permeability (mD)

* $\phi$ is porosity (fraction)

* 0.0314 is a unit conversion factor when permeability is in millidarcies

Higher Rock Quality Index values indicate greater pore connectivity and improved fluid flow capacity.

The Rock Quality Index is commonly used to:

* compare rock quality between reservoirs or geological facies

* identify hydraulic flow units

* calculate the Flow Zone Indicator (FZI)

The Rock Quality Index normalizes permeability by porosity, reducing the influence of pore volume alone and emphasizing pore connectivity.

#### **Sampling Distribution**

A sampling distribution is the probability distribution of a [statistic](GeostatsPy_glossary_n_s.ipynb#statistic) calculated from repeated samples drawn from the same population.

For example, the sampling distribution of the sample mean describes the expected variation in:

$$
\bar{X}
$$

over many possible samples of size $n$ from a population.

The sampling distribution depends on:

* the population distribution

* the sample size

* the statistic being calculated

Sampling distributions are fundamental to statistical inference because they describe the expected variability due to random sampling.

The [standard error](GeostatsPy_glossary_n_s.ipynb#standard-error) is the standard deviation of a sampling distribution, for example,

$$
SE=\sigma_{\bar{X}}
$$

where $\sigma_{\bar{X}}$ is the standard deviation of the sampling distribution of the sample mean.

The [Central Limit Theorem](GeostatsPy_glossary_a_d.ipynb#central-limit-theorem) states that the sampling distribution of the sample mean approaches a [Gaussian distribution](GeostatsPy_glossary_e_m.ipynb#gaussian-distribution) as sample size increases, even when the original population distribution is not Gaussian.

#### **Saturation**

Saturation is the fraction of pore volume occupied by a specific fluid phase.

Fluid saturation is calculated as:

$$
S_i=\frac{V_i}{V_p}
$$

where:
* $S_i$ is the saturation of fluid phase $i$

* $V_i$ is the volume of fluid phase $i$

* $V_p$ is pore volume

Common fluid saturations include:

* water saturation, $S_w$

* oil saturation, $S_o$

* gas saturation, $S_g$

The saturations sum to one:

$$
S_w+S_o+S_g=1
$$

Saturation is a fundamental reservoir property used to estimate hydrocarbon volumes, fluid flow, and production performance.

#### **Score**

A score is a standardized measure that quantifies how far an observation or estimate lies from a reference value relative to a measure of uncertainty or variability.

Practical rules of thumb to determine a small sample size:

* for estimating a population mean, a sample size of approximately $n < 30$ is often considered a small sample, and Student's [t-score](GeostatsPy_glossary_t_z.ipynb#t-score) is typically preferred over the [z-score](GeostatsPy_glossary_t_z.ipynb#z-score) when the population standard deviation is unknown.

* for estimating a population proportion, the normal approximation is generally appropriate only when both,

$$
np \ge 30
$$

and

$$
n(1-p) \ge 30,
$$

where $p$ is the population proportion. Otherwise, exact or small-sample methods are preferred.

Scores are generally calculated as:

$$
\text{score}=\frac{\text{observation}-\text{reference value}}{\text{measure of variability}}
$$

where the measure of variability depends on the application.

Common examples include:

* [Z-score](GeostatsPy_glossary_t_z.ipynb#z-score) – assuming a large sample size

* [t-score](GeostatsPy_glossary_t_z.ipynb#t-score) – accounting for small sample size

Scores are commonly used to:

* compare observations from different populations or distributions

* calculate probabilities and quantiles

* construct [confidence interval](GeostatsPy_glossary_a_d.ipynb#confidence-interval)s

* perform statistical hypothesis tests

#### **Secondary Data**

Data [sample](GeostatsPy_glossary_n_s.ipynb#sample)s of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) other than the feature being modeled, used to improve [estimation](GeostatsPy_glossary_e_m.ipynb#estimation) or [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) of the primary feature.

Secondary data are integrated through a model of the relationship between the secondary and [primary data](GeostatsPy_glossary_n_s.ipynb#primary-data).

For example,

* acoustic impedance measurements from [seismic](GeostatsPy_glossary_n_s.ipynb#seismic) data (secondary data) are used to support calculation of a 3D porosity model, where porosity is the feature of interest

* porosity measurements (secondary data) are used to support calculation of a permeability model, where permeability is the feature of interest

Secondary data may provide additional spatial information, [trend](GeostatsPy_glossary_t_z.ipynb#trend)s, or constraints, but are not the direct observations of the feature being modeled.

#### **Seismic**

A geophysical measurement technique that uses controlled acoustic sources and receivers to measure subsurface reflections and infer geological structure and rock properties.

Reflection seismic data provide high spatial coverage but generally lower resolution compared with direct measurements such as [well log](GeostatsPy_glossary_t_z.ipynb#well-log) and [core](GeostatsPy_glossary_a_d.ipynb#core) data.

Some important details include,

* seismic reflection amplitudes are processed and inverted to estimate rock properties, such as acoustic impedance, calibrated and positionally aligned with well sonic logs

* seismic provides a geological framework by identifying bounding surfaces, structural features, and reservoir extents

* seismic provides soft information for reservoir properties, such as porosity and [facies](GeostatsPy_glossary_e_m.ipynb#facies), through relationships established between seismic attributes and available [primary data](GeostatsPy_glossary_n_s.ipynb#primary-data)

In geostatistical modeling, seismic is commonly used as [secondary data](GeostatsPy_glossary_n_s.ipynb#secondary-data) to improve spatial [prediction](GeostatsPy_glossary_n_s.ipynb#prediction) and uncertainty models.

#### **Sequential Gaussian Simulation**

A stochastic [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) method used to calculate equiprobable spatial model [realization](GeostatsPy_glossary_n_s.ipynb#realization)s by sequentially sampling from local conditional uncertainty distributions.

The name describes the three fundamental principles,

* Sequential - previously simulated values are sequentially added to the conditioning data set so that the simulated realization honors the modeled spatial [covariance](GeostatsPy_glossary_a_d.ipynb#covariance).

* Gaussian - data are transformed to Gaussian space so that local uncertainty distributions can be calculated using the [kriging](GeostatsPy_glossary_e_m.ipynb#kriging) [mean](GeostatsPy_glossary_e_m.ipynb#mean) and [kriging variance](GeostatsPy_glossary_e_m.ipynb#kriging-variance) under the Gaussian [random function](GeostatsPy_glossary_n_s.ipynb#random-function) assumption. After simulation, values are back-transformed to reproduce the original feature distribution.

* Simulation - local conditional distributions are sampled using [Monte Carlo simulation](GeostatsPy_glossary_e_m.ipynb#monte-carlo-simulation) to add the missing spatial variability and calculate multiple equiprobable realizations. The random seed controls both the local random draws and the sequential simulation path.

The complete sequential Gaussian simulation workflow is,

1. Establish the simulation grid and coordinate system, including geological framework transformations such as flattening folds and restoring faults.

2. Assign available data to the simulation grid, accounting for scale changes between data support and grid cell support.

3. Transform the data to Gaussian space using Gaussian anamorphosis.

4. Calculate and model the [variogram](GeostatsPy_glossary_t_z.ipynb#variogram) of the Gaussian transformed data.

5. Establish a random simulation path through all grid nodes. At each node,

    * identify nearby conditioning data and previously simulated grid nodes

    * calculate the local conditional distribution using kriging, where the mean is the kriging estimate and the [variance](GeostatsPy_glossary_t_z.ipynb#variance) is the kriging variance

    * Monte Carlo simulate a realization from the local conditional distribution

    * add the simulated value to the conditioning data set for subsequent simulation steps

6. Check the realization in Gaussian space. The realization should honor,

    * conditioning data at sampled locations

    * the standard normal distribution, $N[0,1]$, with mean zero and [variance](GeostatsPy_glossary_t_z.ipynb#variance) one

    * the modeled variogram

7. Back-transform simulated values from Gaussian space to the original feature distribution.

8. Restore the original geological framework, including folds and faults.

9. Check that the realization honors,

    * geological concepts

    * geophysical data

    * historical [production data](GeostatsPy_glossary_n_s.ipynb#production-data)

10. Repeat steps 5 through 9 to calculate multiple realizations.

The critical algorithmic steps of sequential Gaussian simulation are,

1. Transform data to Gaussian space with mean 0.0 and variance 1.0 (standard normal).

2. Apply a random simulation path through the grid. At each grid node,

    * use kriging to calculate the local conditional distribution

    * Monte Carlo simulate a local realization

    * add the simulated value to the conditioning data

3. Back-transform the simulated values to the original feature distribution.

#### **Sequential Indicator Simulation**

A stochastic [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) method to calculate [continuous feature](GeostatsPy_glossary_a_d.ipynb#continuous-feature) or [categorical feature](GeostatsPy_glossary_a_d.ipynb#categorical-feature) [realization](GeostatsPy_glossary_n_s.ipynb#realization)s for spatial models based on the following principles,

* Sequential - previously simulated values are sequentially added to the conditioning data set so that the simulated realization honors the modeled spatial [covariance](GeostatsPy_glossary_a_d.ipynb#covariance).

* Indicator - [indicator transform](GeostatsPy_glossary_e_m.ipynb#indicator-transform) is applied to the data over a set of thresholds for continuous features or categories for categorical features. This allows flexible probability encoding of hard and soft data and direct estimation of local [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)s or categorical probability distributions.

* Simulation - [Monte Carlo simulation](GeostatsPy_glossary_e_m.ipynb#monte-carlo-simulation) is applied to the local distributions to add the missing [variance](GeostatsPy_glossary_t_z.ipynb#variance) and construct multiple, equiprobable realizations. The random seed determines the individual Monte Carlo simulations and the random path for the sequential simulation.

The sequential indicator simulation workflow is,

1. Establish the simulation grid and coordinate system, including geological framework transformations such as flattening folds and restoring faults.

2. Assign available data to the simulation grid, accounting for scale changes between data support and grid cell support.

3. Apply the indicator transform to all data for all thresholds or categories, $k = 1,\ldots,K$.

4. Calculate and model the [indicator variogram](GeostatsPy_glossary_e_m.ipynb#indicator-variogram) for each threshold or category, $k = 1,\ldots,K$.

5. Establish a random simulation path through all grid nodes. At each node,

    * identify nearby conditioning data and previously simulated grid nodes

    * for each threshold or category, calculate the local [conditional probability](GeostatsPy_glossary_a_d.ipynb#conditional-probability) distribution using [indicator kriging](GeostatsPy_glossary_e_m.ipynb#indicator-kriging)

    * apply [order relations correction](GeostatsPy_glossary_n_s.ipynb#order-relations-correction) to ensure the local cumulative distribution function or probability distribution is valid

    * Monte Carlo simulate a realization from the local conditional distribution

    * add the simulated value to the conditioning data set and apply the indicator transform for subsequent simulation steps

6. Check the realization. Does it honor,

    * conditioning data at sampled locations?

    * the global [histogram](GeostatsPy_glossary_e_m.ipynb#histogram) or [categorical feature](GeostatsPy_glossary_a_d.ipynb#categorical-feature) [proportions](GeostatsPy_glossary_n_s.ipynb#proportions)?

    * the modeled [indicator variogram](GeostatsPy_glossary_e_m.ipynb#indicator-variogram)s?

7. Restore the original geological framework, including adding back folds and faults.

8. Check that the realization honors,

    * geological concepts

    * geophysical data

    * historical [production data](GeostatsPy_glossary_n_s.ipynb#production-data)

9. Repeat steps 5 through 8 to calculate multiple realizations.

#### **Significance Level**

The significance level is the probability threshold used in [hypothesis testing](GeostatsPy_glossary_e_m.ipynb#hypothesis) to determine whether to reject the [null hypothesis](GeostatsPy_glossary_n_s.ipynb#null-hypothesis).

The significance level is also called the [alpha level](GeostatsPy_glossary_a_d.ipynb#alpha-level):

$$
\alpha
$$

The significance level also determines the corresponding confidence level for [confidence intervals](GeostatsPy_glossary_a_d.ipynb#confidence-interval).

Common choices include:

* $\alpha=0.10$ for 90% confidence

* $\alpha=0.05$ for 95% confidence

* $\alpha=0.01$ for 99% confidence

The significance level defines the probability of incorrectly rejecting a true null hypothesis:

$$
P(\text{Reject }H_0|H_0\text{ is true})=\alpha
$$

This error is known as a [type I error](GeostatsPy_glossary_t_z.ipynb#type-i-error).

The significance level is selected before performing the hypothesis test and represents the acceptable false positive risk.

Selecting a significance level requires balancing sensitivity and false positive risk:

* too small — the test becomes conservative, requiring very strong evidence before identifying an effect; real effects may be missed

* too large — the test becomes permissive, identifying more outcomes as unusual but increasing false positive risk

The fundamental question is:

* How rare must an event be before we consider it evidence against our assumption?

* How often should our process generate a red alert when the process is actually operating normally?

#### **Sill**

For a stationary [random function](GeostatsPy_glossary_n_s.ipynb#random-function), the theoretical [variogram](GeostatsPy_glossary_t_z.ipynb#variogram) sill is equal to the [variance](GeostatsPy_glossary_t_z.ipynb#variance) of the feature of interest.

The sill provides the reference for interpreting [spatial continuity](GeostatsPy_glossary_n_s.ipynb#spatial-continuity),

* at [lag distance](GeostatsPy_glossary_e_m.ipynb#lag-distance)s where the [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram) reaches the sill, spatial correlation approaches zero. This lag distance is called the [range](GeostatsPy_glossary_n_s.ipynb#range).

* [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram) values above the sill may indicate negative [covariance function](GeostatsPy_glossary_a_d.ipynb#covariance-function), but are also commonly caused by sampling variability, insufficient data, or non-[stationarity](GeostatsPy_glossary_n_s.ipynb#stationarity). In practice, variogram models are generally constrained to approach the sill and assume no spatial correlation beyond the range.

* an experimental variogram that continues to increase approximately linearly beyond the sill may indicate a spatial [trend](GeostatsPy_glossary_t_z.ipynb#trend) or other violation of stationarity assumptions.

For a feature with [variance](GeostatsPy_glossary_t_z.ipynb#variance), $\sigma^2$, the variogram sill is,

$$
\gamma(\mathbf{h}) \rightarrow \sigma^2
$$

as the lag distance increases beyond the range.

#### **Simulation**

A stochastic process of obtaining one or more possible values of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) at unsampled locations that are consistent with available data and a spatial uncertainty model.

Simulation models are designed to reproduce global characteristics and spatial variability, known as [Global accuracy](GeostatsPy_glossary_e_m.ipynb#global-accuracy), where the model reproduces specified [global measures](GeostatsPy_glossary_e_m.ipynb#global-measure), including,

* feature [univariate](GeostatsPy_glossary_t_z.ipynb#univariate) distributions, including [proportions](GeostatsPy_glossary_n_s.ipynb#proportions), [mean](GeostatsPy_glossary_e_m.ipynb#mean), variance, and the complete [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)

* feature spatial models, including the [variogram](GeostatsPy_glossary_t_z.ipynb#variogram), [training image](GeostatsPy_glossary_t_z.ipynb#training-image), or geological object geometries

Unlike [estimation](GeostatsPy_glossary_e_m.ipynb#estimation) methods, simulation produces multiple equiprobable [realization](GeostatsPy_glossary_n_s.ipynb#realization)s that represent uncertainty rather than a single optimal [prediction](GeostatsPy_glossary_n_s.ipynb#prediction).

Examples of simulation models include,

* subsurface heterogeneity models, including [sequential Gaussian simulation](GeostatsPy_glossary_n_s.ipynb#sequential-gaussian-simulation), [sequential indicator simulation](GeostatsPy_glossary_n_s.ipynb#sequential-indicator-simulation), [multiple point simulation](GeostatsPy_glossary_e_m.ipynb#multiple-point-simulation), and [object-based simulation](GeostatsPy_glossary_n_s.ipynb#object-based-simulation)

* uncertainty propagation through a [transfer function](GeostatsPy_glossary_t_z.ipynb#transfer-function), including [Monte Carlo simulation](GeostatsPy_glossary_e_m.ipynb#monte-carlo-simulation)

Use simulation when,

* reproducing feature distributions is important, especially when extreme values influence decisions

* realistic spatial models are required for applications such as flow simulation

* uncertainty in the [decision criteria](GeostatsPy_glossary_a_d.ipynb#decision-criteria) must be quantified through multiple possible models

#### **Simulation Post-processing**

A variety of operations to calculate statistical summaries over multiple [realizations](GeostatsPy_glossary_n_s.ipynb#realizations) and [scenarios](GeostatsPy_glossary_n_s.ipynb#scenarios). At each location in the model, $\bf{u}_{\alpha}$, we pool the local realizations, $\ell = 1, \ldots, L$,

$$
\left\{x^{\ell}(\mathbf{u}_{\alpha}): \ell = 1,\ldots,L \right\}$$

and then calculate the nonparametric local [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function) as,

$$
F_x(x)=P(X\le x)
$$

Here are common summaries and how they are used,

1. e-type - is the local expectation (since each realization is assumed to be equiprobable, this is the same as the local [average](GeostatsPy_glossary_a_d.ipynb#average)) over the realizations,

$$
e(\bf{u}*{\alpha}) = E \left[x^{\ell}(\bf{u}*{\alpha})\right], \quad \forall \ell = 1,\ldots,L,\quad \alpha \in V
$$

* often calculated to visualize the [trend](GeostatsPy_glossary_t_z.ipynb#trend)s after integration of all information sources

2. Conditional standard deviation - is the local standard deviation over the realizations,

$$
\sigma(\bf{u}*{\alpha}) = \sqrt{E \left[(x^{\ell}(\bf{u}*{\alpha})-e(\bf{u}_{\alpha}))^2\right]}, \quad \forall \ell = 1,\ldots,L,\quad \alpha \in V
$$

* often calculated to visualize the level of local uncertainty

3. Local percentile - is the local percentile over the realizations,

$$
x^p(\bf{u}*{\alpha}) = F^{-1}(p;\bf{u}*{\alpha}), \quad \alpha \in V
$$

* often calculated to show maps of local upper and lower bounds, for example, local P10 and P90 models

4. Local probability of exceedance - the probability of exceeding a specified threshold, $z_k$, calculated as one minus the cumulative probability of the threshold at each location over the realizations,

$$
P{X(\bf{u}*{\alpha}) \le x_k} = 1 - \frac{1}{L}\cdot\sum*{\ell=1}^{L}i^{\ell}(\bf{u}_{\alpha};z_k), \quad \alpha \in V
$$

* often calculated to communicate risk, for example, the probability of locally exceeding an environmental threshold for a contaminant

A sufficient number of realizations, $L$, is required to calculate reliable summaries,

* the expected value is the easiest to calculate and is generally reliable with 20 or more realizations

* percentiles on the tails, e.g., P10 and P90, require 100 or more realizations for reliable results

More realizations generally result in more reliable simulation post-processing and improved decision support.

- [Simulation Postprocessing Description and Demonstration](GeostatsPy_simulation_postsim.ipynb#local-uncertainty)


#### **Skewness**

Skewness is a [univariate parameter](GeostatsPy_glossary_t_z.ipynb#univariate-parameter) or [univariate statistic](GeostatsPy_glossary_t_z.ipynb#univariate-statistic) that measures the asymmetry of the probability distribution of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) or [variable](GeostatsPy_glossary_t_z.ipynb#variable).

A common measure of sample skewness is:

$$
Skewness = \frac{\frac{1}{n}\sum_{i=1}^{n}(x_i-\bar{x})^3}{\left(\frac{1}{n}\sum_{i=1}^{n}(x_i-\bar{x})^2\right)^{3/2}}
$$

Interpretation:

* positive skewness indicates a longer upper tail with occasional large values

* negative skewness indicates a longer lower tail with occasional small values

* zero skewness indicates a symmetric distribution

Skewness is important for subsurface variables because many geological and engineering properties are not symmetric and may contain extreme values.

#### **Soft Data**

#### **Spatial Estimation**

The process of obtaining a single best value to represent a [feature](GeostatsPy_glossary_e_m.ipynb#feature) at an unsampled location or time, $\bf{u}$.

Given spatial data, $z(\bf{u}_1), \dots, z(\bf{u}_n)$, we estimate the unknown feature value at location $\bf{u}$ with a linear combination of the available data,

$$
z^{*}(\bf{u}) = \sum_{\alpha=1}^{n} \lambda_{\alpha} z(\bf{u}_{\alpha})
$$

An unbiasedness constraint may be added by assigning the remainder of the weight (one minus the sum of weights) to the global [average](GeostatsPy_glossary_a_d.ipynb#average). Therefore, if no informative data are available, the estimate approaches the global average of the feature,

$$
z^{*}(\bf{u}) = \sum_{\alpha=1}^{n} \lambda_{\alpha} z(\bf{u}*{\alpha}) + \left(1-\sum*{\alpha=1}^{n}\lambda_{\alpha}\right)\overline{z}
$$

Some additional concepts,

* local accuracy takes precedence over [global accuracy](GeostatsPy_glossary_e_m.ipynb#global-accuracy), meaning that spatial estimation methods prioritize matching nearby observations over reproducing global statistics such as the [histogram](GeostatsPy_glossary_e_m.ipynb#histogram) and variogram

* spatial estimation maps and models generally have reduced [variance](GeostatsPy_glossary_t_z.ipynb#variance) and increased [spatial continuity](GeostatsPy_glossary_n_s.ipynb#spatial-continuity), resulting in smoother models than the true heterogeneous feature distribution

* [estimation](GeostatsPy_glossary_e_m.ipynb#estimation) models are not appropriate for [transfer function](GeostatsPy_glossary_t_z.ipynb#transfer-function)s that are sensitive to heterogeneity and feature distributions, such as flow simulation or economic optimization

* spatial estimation produces a single [deterministic model](GeostatsPy_glossary_a_d.ipynb#deterministic-model) and therefore does not provide multiple [realization](GeostatsPy_glossary_n_s.ipynb#realization)s required to sample uncertainty in the [decision criteria](GeostatsPy_glossary_a_d.ipynb#decision-criteria); [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) methods are required for comprehensive [uncertainty modeling](GeostatsPy_glossary_t_z.ipynb#uncertainty-modeling) and decision support

Examples of spatial estimation methods include,

* inverse distance

* [kriging](GeostatsPy_glossary_e_m.ipynb#kriging)

* k-nearest neighbors

#### **Spatial Continuity**

Also known as spatial correlation, is the correlation of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) over distance.

Spatial continuity can be calculated from a [variogram](GeostatsPy_glossary_t_z.ipynb#variogram). The [correlogram](GeostatsPy_glossary_a_d.ipynb#correlogram), $\rho(\bf{h})$, is related to the variogram, $\gamma(\bf{h})$,

$$
\rho(\bf{h}) = 1.0 - \frac{\gamma(\bf{h})}{\sigma^2}
$$

where $\rho(\bf{h})$ is the spatial correlation at [lag](GeostatsPy_glossary_e_m.ipynb#lag) vector $\bf{h}$, and $\sigma^2$ is the feature variance, the variogram [sill](GeostatsPy_glossary_n_s.ipynb#sill).

* the greater the difference between the variogram value and the sill, the greater the spatial continuity

* at the sill, the variogram indicates no correlation over that separation distance and the correlogram approaches zero

Spatial continuity may be considered,

* between [sample](GeostatsPy_glossary_n_s.ipynb#sample) data during calculation of an [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram)

* while inferring a [variogram model](GeostatsPy_glossary_t_z.ipynb#variogram-model), including selecting nested structures, contributions, and [range](GeostatsPy_glossary_n_s.ipynb#range)s

* between simulated values during construction of a spatial [simulation](GeostatsPy_glossary_n_s.ipynb#simulation) model

How do we interpret spatial continuity?

* no spatial continuity - values are uncorrelated over distance, meaning knowing a value at one location provides no information about the value at another location

* homogeneous phenomena - values are identical or nearly identical over space and therefore exhibit perfect conceptual continuity over all distances; note that a perfectly homogeneous feature has zero variance, so the correlogram formulation is not defined

* strong spatial continuity - nearby values are highly correlated, and the variogram increases slowly with [lag distance](GeostatsPy_glossary_e_m.ipynb#lag-distance)

* weak spatial continuity - nearby values have limited correlation, and the variogram approaches the sill over short distances

Used in:
- [Variogram Calculation Discussion and Demonstration](GeostatsPy_variogram_calculation.ipynb#spatial-continuity)


#### **Standard Deviation**

The square root of the [variance](GeostatsPy_glossary_t_z.ipynb#variance). Standard deviation measures the spread of a feature about its [average](GeostatsPy_glossary_a_d.ipynb#average) in the same units as the original feature.

Given the sample variance, 

$$
s^2 = \frac{1}{n-1}\sum_{\alpha=1}^{n}\left(x_{\alpha}-\overline{x}\right)^2
$$

the sample standard deviation is,

$$
s = \sqrt{s^2}
$$

The equivalent [population](GeostatsPy_glossary_n_s.ipynb#population) [parameter](GeostatsPy_glossary_n_s.ipynb#parameter)s are, the population variance,

$$
\sigma^2 = \frac{1}{n}\sum_{\alpha=1}^{n}\left(x_{\alpha}-\overline{x}\right)^2
$$

the population standard deviation is,

$$
\sigma = \sqrt{\sigma^2}
$$

#### **Standard Error**

The standard error (SE) is the estimated standard deviation of a sample statistic, most commonly the sample [mean](GeostatsPy_glossary_e_m.ipynb#mean). It quantifies the uncertainty associated with estimating a population parameter from a sample.

For the sample mean, the standard error is:

$$
SE=\frac{s}{\sqrt{n}}
$$

where:

* $s$ is the sample [standard deviation](GeostatsPy_glossary_n_s.ipynb#standard-deviation)

* $n$ is the sample size

As sample size increases, the standard error decreases, indicating greater confidence in the estimated population parameter.

Standard error should not be confused with standard deviation, which measures the variability of the data rather than the uncertainty of a statistic estimated from the data.

#### **Stationarity**

The decision that a subset of the subsurface is the same "stuff" and therefore can be pooled to calculate statistics and build models.

Replicates are required to calculate any statistic. In many applications, replicates are obtained by repeated measurements through time, for example,

* air or water [sample](GeostatsPy_glossary_n_s.ipynb#sample)s collected repeatedly from a monitoring station

For subsurface resource models,

* repeated samples are generally not available at the same location; only one sample is available at each location

* instead of pooling measurements through time, we must pool samples over space to calculate statistics

Why must we pool data? Ultimately, it is required to make [inference](GeostatsPy_glossary_e_m.ipynb#inference) about the [population](GeostatsPy_glossary_n_s.ipynb#population) from a limited [sample](GeostatsPy_glossary_n_s.ipynb#sample),

* to calculate statistics

* to build spatial models

The choice of stationary domain is an expert geological decision. Without a stationarity decision, we are restricted to the measured locations (well bores or drill holes) and cannot calculate statistics or make [prediction](GeostatsPy_glossary_n_s.ipynb#prediction)s between samples.

An example geological definition of stationarity could be:

*The rock within the stationary domain is sourced, deposited, preserved, and post-depositionally altered in a similar manner. The domain is mappable and may be used for local prediction or as information for analogous locations within the subsurface; therefore, information may be pooled over this expert-defined volume of the subsurface.*

This expert geological interpretation defines a domain over which statistical stationarity is assumed for modeling.

There are two aspects of any stationarity decision:

1. Import license - the choice of which samples are allowed to contribute to the calculation of a statistic

2. Export license - the choice of where the resulting statistic is applicable within the subsurface

To state a stationarity decision, we must specify:

1. the statistic assumed stationary, for example, the mean, variance, [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function) , or spatial continuity

2. the spatial domain over which the statistic is assumed stationary, for example, the entire model, a [facies](GeostatsPy_glossary_e_m.ipynb#facies), a depositional environment, or a geological region

Examples of statistical definitions of stationarity include:

* stationary mean

$$
E[Z(\mathbf{u})] = \overline{z}, \quad \forall \mathbf{u} \in AOI
$$

* stationary cumulative distribution function

$$
F_z(\mathbf{u},z) = F_z(z), \quad \forall \mathbf{u} \in AOI
$$

* stationary semivariogram

$$
\gamma_z(\mathbf{u},\mathbf{h}) = \gamma_z(\mathbf{h})
$$

The stationarity decision may be extended to any statistic of interest, including,

* [facies](GeostatsPy_glossary_e_m.ipynb#facies) [proportions](GeostatsPy_glossary_n_s.ipynb#proportions)

* [bivariate](GeostatsPy_glossary_a_d.ipynb#bivariate) distributions

* multiple point statistics

Additional considerations for stationarity include:

* Stationarity is a decision, not a hypothesis - therefore, it is not directly tested. Instead, data may demonstrate that a chosen stationarity decision is inappropriate.

* Stationarity depends on scale - the appropriate modeling scale should be selected based on the geological process, decision objective, and project requirements.

* A stationarity decision cannot be avoided - without stationarity, spatial statistics cannot be calculated and modeling cannot progress beyond measured locations. Conversely, assuming broad stationarity over very large regions of the Earth is generally unrealistic.

* Geomodeling stationarity is a domain decision - defining (1) where data may be pooled (import license) and (2) where resulting statistics may be applied (export license).

* Nonstationary trends may be modeled explicitly - deterministic [trend](GeostatsPy_glossary_t_z.ipynb#trend)s can be removed and the remaining stationary residual variation can be modeled stochastically. This is the [hybrid modeling](GeostatsPy_glossary_e_m.ipynb#hybrid-model) approach.

Used in:
- [Variogram Calculation Discussion and Demonstration](GeostatsPy_variogram_calculation.ipynb#stationarity)


#### **Statistic**

A function of sample data that summarizes a property of the [sample](GeostatsPy_glossary_n_s.ipynb#sample). Examples include,

* sample [mean](GeostatsPy_glossary_e_m.ipynb#mean) - $\overline{x}$

* sample [standard deviation](GeostatsPy_glossary_n_s.ipynb#standard-deviation) - $s$

* sample [variance](GeostatsPy_glossary_t_z.ipynb#variance) - $s^2$

* [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)  - $F_x(x)$

* [experimental variogram](GeostatsPy_glossary_e_m.ipynb#experimental-variogram) - $\gamma(\mathbf{h})$

Statistics are calculated from available samples because the complete [population](GeostatsPy_glossary_n_s.ipynb#population) is generally unknown.

How do we use statistics?

* [Inference](GeostatsPy_glossary_e_m.ipynb#inference) - calculate [statistics](GeostatsPy_glossary_n_s.ipynb#statistic) from a [sample](GeostatsPy_glossary_n_s.ipynb#sample) and use them to estimate unknown population [parameter](GeostatsPy_glossary_n_s.ipynb#parameter)s

* [Prediction](GeostatsPy_glossary_n_s.ipynb#prediction) - use a model of population [parameter](GeostatsPy_glossary_n_s.ipynb#parameter)s and relationships to predict future observations or outcomes

#### **Statistical Distribution**

A description of the frequency or probability behavior of a [feature](GeostatsPy_glossary_e_m.ipynb#feature) over the range of possible values.

A [univariate](GeostatsPy_glossary_t_z.ipynb#univariate) [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) describes how feature values are distributed without considering their spatial or temporal arrangement. We represent the statistical distribution with,

* [histogram](GeostatsPy_glossary_e_m.ipynb#histogram)

* [normalized histogram](GeostatsPy_glossary_n_s.ipynb#normalized-histogram)

* [probability density function](GeostatsPy_glossary_n_s.ipynb#probability-density-function)

* [cumulative distribution function](GeostatsPy_glossary_a_d.ipynb#cumulative-distribution-function)

What do we learn from a statistical distribution? For example,

* what are the minimum and maximum values?

* what is the most common range of values?

* do we have many low values?

* do we have many high values?

* are there outliers or values that do not make geological or physical sense and require explanation?

* what is the variability and uncertainty in the feature values?

Statistical distributions are fundamental for [inference](GeostatsPy_glossary_e_m.ipynb#inference), [simulation](GeostatsPy_glossary_n_s.ipynb#simulation), and [uncertainty modeling](GeostatsPy_glossary_t_z.ipynb#uncertainty-modeling).

#### **Statistical Power**

Statistical power is the probability that a [hypothesis test](GeostatsPy_glossary_e_m.ipynb#hypothesis) correctly rejects the [null hypothesis](GeostatsPy_glossary_n_s.ipynb#null-hypothesis) when a real effect exists.

Statistical power is defined as:

$$
Power=1-\beta
$$

where:

* $\beta$ is the probability of a [Type II Error](GeostatsPy_glossary_t_z.ipynb#type-ii-error)

Higher statistical power increases the probability of detecting meaningful effects.

Statistical power increases with:

* larger sample size

* lower variability

* larger effect size

* appropriate selection of significance level

Low statistical power may result in failing to detect real differences or relationships.

- [Hypothesis Testing Definition and Demonstration](GeostatsPy_hypothesis_testing.ipynb#hypothesis-testing-concepts)

#### **Stochastic Model**

A model of a system or process that includes uncertainty and is represented by multiple possible outcomes, including [realizations](GeostatsPy_glossary_n_s.ipynb#realizations) and [scenarios](GeostatsPy_glossary_n_s.ipynb#scenarios), constrained by available data, statistics, and modeling assumptions.

Stochastic models represent uncertainty by describing a range of plausible outcomes rather than a single deterministic [prediction](GeostatsPy_glossary_n_s.ipynb#prediction).

Examples include,

* data-driven models that integrate uncertainty, such as geostatistical simulation models

* Monte Carlo models that propagate uncertainty through a [transfer function](GeostatsPy_glossary_t_z.ipynb#transfer-function)

* ensemble machine learning models that represent prediction uncertainty

Advantages:

* computational speed compared with many physics-based models

* explicit uncertainty assessment

* ability to report confidence intervals, prediction intervals, and risk measures

* ability to integrate many sources of data and information

* flexible data-driven approaches

Disadvantages:

* limited representation of underlying physics unless explicitly incorporated

* dependence on statistical model assumptions and simplifications

* uncertainty models may be incomplete if important processes or information are not represented

Used in:
- [Comparison of Deterministic and Stochastic Modeling](GeostatsPy_concepts.ipynb#deterministic-vs-stochastic-modeling)

Contrast with:
- [Deterministic Model](GeostatsPy_glossary_a_d.ipynb#deterministic-model)


#### **Student's t Distribution**

Student's t distribution is a continuous [statistical distribution](GeostatsPy_glossary_n_s.ipynb#statistical-distribution) used for inference when estimating a population mean with unknown population [standard deviation](GeostatsPy_glossary_n_s.ipynb#standard-deviation).

The t distribution is defined by its degrees of freedom, $\nu$.

A t-score is calculated as:

$$
t=\frac{\bar{x}-\mu_0}{s/\sqrt{n}}
$$

where:

* $\bar{x}$ is the sample mean

* $\mu_0$ is the hypothesized population mean

* $s$ is the sample standard deviation

* $n$ is the number of samples

The t distribution accounts for additional uncertainty from estimating the population standard deviation using the sample standard deviation.

Important characteristics:

* symmetric around zero

* heavier tails than the [Gaussian distribution](GeostatsPy_glossary_e_m.ipynb#gaussian-distribution)

* approaches the Gaussian distribution as degrees of freedom increase

The t distribution is commonly used for:

* small sample inference

* confidence intervals

* hypothesis tests involving sample means

#### **Subsurface Modeling Workflow**

A common geostatistical workflow for integrating subsurface data, modeling uncertainty, and supporting development [decision making](GeostatsPy_glossary_a_d.ipynb#decision-making). The workflow proceeds from data to decisions through the following steps:

1. Integrate all available information to build multiple subsurface [scenarios](GeostatsPy_glossary_n_s.ipynb#scenarios) and [realizations](GeostatsPy_glossary_n_s.ipynb#realizations) that sample the uncertainty space.

2. Apply all realizations through the [transfer function](GeostatsPy_glossary_t_z.ipynb#transfer-function) to sample the uncertainty in the decision criteria.

3. Assemble the distribution of the [decision criteria](GeostatsPy_glossary_a_d.ipynb#decision-criteria) from the ensemble of realizations and scenarios.

4. Make the optimum reservoir development decisions while accounting for the modeled uncertainty.

Used in:
- [The Common Geostatistical Modeling Workflow](GeostatsPy_concepts.ipynb#the-common-geostatistical-modeling-workflow)
